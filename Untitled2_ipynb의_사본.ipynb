```python
import pandas as pd

df = pd.read_csv('/content/전처리 완료.csv', encoding='cp949')
display(df)
```

```python
# Separate the target variable (y) from the features (X)
y = df['고위험군여부']
X = df.drop('고위험군여부', axis=1)

print("Features (X) head:")
display(X.head())

print("\nTarget (y) head:")
display(y.head())
```

```python
# Install imbalanced-learn if you haven't already
# !pip install imbalanced-learn
```

```python
from imblearn.under_sampling import RandomUnderSampler

# Instantiate RandomUnderSampler
# By default, it undersamples the majority class to make its count equal to the minority class
undersampler = RandomUnderSampler(random_state=42)

# Apply undersampling to X and y
X_resampled, y_resampled = undersampler.fit_resample(X, y)

print("Original target distribution:")
display(y.value_counts())

print("\nResampled target distribution:")
display(y_resampled.value_counts())
```

```python
from sklearn.model_selection import train_test_split

# Split the resampled data into training and testing sets
# test_size=0.2 means 20% of the data will be used for testing
# random_state for reproducibility
X_train, X_test, y_train, y_test = train_test_split(X_resampled, y_resampled, test_size=0.2, random_state=42, stratify=y_resampled)

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")

print("\nClass distribution in y_train:")
display(y_train.value_counts())

print("\nClass distribution in y_test:")
display(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Initialize the StandardScaler
scaler = StandardScaler()

# Fit the scaler on the training data and transform it
X_train_scaled = scaler.fit_transform(X_train)

# Transform the test data using the fitted scaler
X_test_scaled = scaler.transform(X_test)

# Display the first 5 rows of the scaled training data to verify
print("First 5 rows of X_train_scaled:")
display(pd.DataFrame(X_train_scaled, columns=X_train.columns).head())

print("\nFirst 5 rows of X_test_scaled:")
display(pd.DataFrame(X_test_scaled, columns=X_test.columns).head())
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Initialize the Logistic Regression model
log_reg_model = LogisticRegression(random_state=42, solver='liblinear') # 'liblinear' is a good choice for small datasets and binary classification

# Train the model using the scaled training data
log_reg_model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully.")
```

```python
# Make predictions on the scaled test data
y_pred = log_reg_model.predict(X_test_scaled)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred)

print(f"\nAccuracy: {accuracy:.4f}")
print("\nConfusion Matrix:")
display(pd.DataFrame(conf_matrix, index=['Actual 0', 'Actual 1'], columns=['Predicted 0', 'Predicted 1']))
print("\nClassification Report:")
print(class_report)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV
import matplotlib.pyplot as plt
import numpy as np

# Define the range of K values to search
k_range = list(range(1, 31))
param_grid = dict(n_neighbors=k_range)

# Initialize KNN classifier
kNN_classifier = KNeighborsClassifier()

# Setup GridSearchCV
grid = GridSearchCV(kNN_classifier, param_grid, cv=10, scoring='accuracy', return_train_score=False, verbose=1, n_jobs=-1)

# Fit GridSearchCV to the scaled training data
grid.fit(X_train_scaled, y_train)

print("GridSearchCV completed.")
```

```python
# Print the best K value and its corresponding accuracy
print(f"Best K: {grid.best_params_['n_neighbors']}")
print(f"Best accuracy: {grid.best_score_:.4f}")

# Get all K values and their mean test scores
k_values = [d['n_neighbors'] for d in grid.cv_results_['params']]
mean_test_scores = grid.cv_results_['mean_test_score']

# Plotting the accuracy for different K values
fig = plt.figure(figsize=(10, 6))
plt.plot(k_values, mean_test_scores, marker='o')
plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Cross-Validated Accuracy')
plt.title('KNN Classifier Accuracy for Different K Values')
plt.xticks(k_values)
plt.grid(True)
plt.show()
```
